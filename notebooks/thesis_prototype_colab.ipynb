## Display Framework Outputs

```python
import json
from pathlib import Path

output_dir = Path("outputs")
for file_name in ["langgraph_output.json", "crewai_output.json", "autogen_output.json"]:
    path = output_dir / file_name
    print(f"\n=== {file_name} ===")
    data = json.loads(path.read_text(encoding="utf-8"))
    print("Framework:", data["framework"])
    print("Status:", data["status"])
    print("Errors:", data["execution_metadata"].get("errors", []))
    print("Title:", data["final_answer_json"].get("title", ""))
```
